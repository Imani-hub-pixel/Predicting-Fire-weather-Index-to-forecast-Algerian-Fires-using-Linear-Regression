```python
import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
algerian_fires=pd.read_csv("Algerian_forest_fires_dataset_UPDATE.csv",header=1)
algerian_fires.head(5)
```

## Data Cleaning

```python
algerian_fires.info()
```

```python
algerian_fires.isnull().sum()
```

```python
algerian_fires[algerian_fires.isnull().any(axis=1)]
```

```python
algerian_fires.loc[:121,"Region"]=0
algerian_fires.loc[122:,"Region"]=1
df=algerian_fires
```

```python
df.info()
```

```python
df[["Region"]]=df[["Region"]].astype(int)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df=df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df.loc[[122]]
```

```python
df=df.drop(122).reset_index(drop=True)
```

```python
df.columns
```

```python
## Remove empty spaces in the column names
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

```python
df[["day","month","year","Temperature","RH","Ws"]]=df[["day","month","year","Temperature","RH","Ws"]].astype(int)
```

```python
df.info()
```

```python
df[["Rain","FFMC","DMC","DC","ISI","BUI","FWI"]]=df[["Rain","FFMC","DMC","DC","ISI","BUI","FWI"]].astype(float)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.to_csv('Algerian_forest_fires_cleaned_dataset.csv',index=False)
```

## Exploratory Data Analysis

```python
df_copy=df.drop(["day","month","year"],axis=1)
```

```python
df_copy.head()
```

```python
df_copy['Classes'].value_counts()
```

```python
df_copy['Classes']=df_copy['Classes'].str.strip()
```

```python
df_copy['Classes'].value_counts()
```

```python
df_copy['Classes']=np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df_copy.head()
```

```python
sns.pairplot(df_copy)
plt.show()
```

```python
sns.set_theme()
df_copy.hist(bins=50,figsize=(20,15))
plt.show()
```

```python
percent=df_copy['Classes'].value_counts(normalize=True)*100
```

```python
labels=["Fire","Not Fire"]
plt.figure(figsize=(12,7))
plt.pie(percent,labels=labels,autopct='%1.1f%%')
plt.title("Pie Chart of Classes")
plt.show()
```

```python
df['Classes']=np.where(df["Classes"].str.contains('not fire'),'not fire','fire')
```

```python
dftemp=df.loc[df["Region"]==1]
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',hue='Classes',data=dftemp)
plt.ylabel("Number rof fires",weight="bold")
plt.xlabel("Months",weight="bold")
plt.title("Fire analysis of Sidi -Bel Regions",weight="bold")
plt.show()
```

```python
dftemp=df.loc[df["Region"]==0]
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',hue='Classes',data=dftemp)
plt.ylabel("Number rof fires",weight="bold")
plt.xlabel("Months",weight="bold")
plt.title("Fire analysis of Brjaia Regions",weight="bold")
plt.show()
```
